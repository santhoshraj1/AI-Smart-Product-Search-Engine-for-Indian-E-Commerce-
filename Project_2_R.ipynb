```python
# Combine all text columns into one search column
df['search_text'] = (
    df['product_name'] + ' ' +
    df['brand'] + ' ' +
    df['category'] + ' ' +
    df['subcategory'] + ' ' +
    df['description']
)

print("Combined text column created!")
print("\nSample search text:")
print(df['search_text'][0])
```

```python
def clean_text(text):
    # Convert to lowercase
    text = text.lower()
    
    # Remove special characters and numbers
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    
    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()
    
    # Remove stopwords
    words = text.split()
    words = [w for w in words if w not in stop_words]
    
    return ' '.join(words)

# Apply cleaning to search text
df['clean_text'] = df['search_text'].apply(clean_text)

print("Text cleaned!")
print("\nBefore cleaning:")
print(df['search_text'][0])
print("\nAfter cleaning:")
print(df['clean_text'][0])
```

```python
# Make sure price columns are numeric
df['price'] = pd.to_numeric(df['price'], errors='coerce')
df['discounted_price'] = pd.to_numeric(df['discounted_price'], errors='coerce')
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

# Fill any missing values
df['price'].fillna(df['price'].mean(), inplace=True)
df['discounted_price'].fillna(df['discounted_price'].mean(), inplace=True)
df['rating'].fillna(0, inplace=True)

print("Price columns cleaned!")
print("\nSample cleaned data:")
df[['product_name', 'price', 'discounted_price', 'rating', 'clean_text']].head()
```

# Build TF-IDF Search Engine 🔍

TF  = Term Frequency
      How many times a word appears in a product

IDF = Inverse Document Frequency
      How unique/rare that word is across all products

TF-IDF = Words that are important to THAT product
         but not common in ALL products

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Build TF-IDF vectorizer
vectorizer = TfidfVectorizer(
    max_features=5000,    # use top 5000 important words
    ngram_range=(1, 2),   # single words + two word phrases
    min_df=1              # include words that appear at least once
)

# Fit and transform the clean text
tfidf_matrix = vectorizer.fit_transform(df['clean_text'])

print("TF-IDF Matrix Built!")
print("Matrix Shape:", tfidf_matrix.shape)
print(f"  → {tfidf_matrix.shape[0]} products")
print(f"  → {tfidf_matrix.shape[1]} unique words/phrases")
```

```python
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))

def clean_query(query):
    # Clean the user query same way as products
    query = query.lower()
    query = re.sub(r'[^a-zA-Z\s]', '', query)
    words = query.split()
    words = [w for w in words if w not in stop_words]
    return ' '.join(words)

def search_products(query, top_n=5):
    
    # Step 1: Clean the query
    cleaned_query = clean_query(query)
    print(f"🔍 Searching for: '{query}'")
    print(f"   Cleaned query: '{cleaned_query}'\n")
    
    # Step 2: Convert query to TF-IDF vector
    query_vector = vectorizer.transform([cleaned_query])
    
    # Step 3: Calculate similarity with all products
    similarity_scores = cosine_similarity(query_vector, tfidf_matrix).flatten()
    
    # Step 4: Get top N most similar products
    top_indices = similarity_scores.argsort()[::-1][:top_n]
    
    # Step 5: Return results
    results = df.iloc[top_indices].copy()
    results['similarity_score'] = similarity_scores[top_indices]
    
    return results[['product_name', 'brand', 'category',
                     'subcategory', 'price', 'discounted_price',
                     'rating', 'stock_status', 'similarity_score']]

print("Search function ready!")
```

### Test with Price Filter

```python
def search_with_price(query, max_price=None, min_price=None, top_n=5):
    
    # Get TF-IDF results
    cleaned_query = clean_query(query)
    query_vector = vectorizer.transform([cleaned_query])
    similarity_scores = cosine_similarity(query_vector, tfidf_matrix).flatten()
    
    # Add scores to dataframe
    temp_df = df.copy()
    temp_df['similarity_score'] = similarity_scores
    
    # Apply price filter
    if max_price:
        temp_df = temp_df[temp_df['discounted_price'] <= max_price]
    if min_price:
        temp_df = temp_df[temp_df['discounted_price'] >= min_price]
    
    # Sort by similarity score
    temp_df = temp_df.sort_values('similarity_score', ascending=False).head(top_n)
    
    return temp_df[['product_name', 'brand', 'category',
                     'discounted_price', 'rating',
                     'stock_status', 'similarity_score']]

# Test price filter
print("=" * 50)
print("TEST 2: Budget phone under ₹20000")
print("=" * 50)
results = search_with_price("smartphone", max_price=20000, top_n=5)
print(results.to_string())
```

```python
def ai_rank_products(query, search_results):
    
    # Build product list for AI
    products_text = ""
    for i, (_, row) in enumerate(search_results.iterrows(), 1):
        products_text += f"""
        Product {i}:
        - Name        : {row['product_name']}
        - Brand       : {row['brand']}
        - Category    : {row['category']}
        - Price       : ₹{row['discounted_price']:,.0f}
        - Rating      : {row['rating']} ⭐
        - Stock       : {row['stock_status']}
        - Match Score : {row['similarity_score']:.2f}
        """
    
    prompt = f"""
    You are a smart Indian e-commerce product search assistant.
    
    Customer searched for: "{query}"
    
    Here are the top matching products found:
    {products_text}
    
    Your task:
    1. 🏆 Rank these products from best to worst match for the query
    2. 💰 Highlight the best value for money product
    3. ⭐ Highlight the highest rated product
    4. 💡 Give a one line buying tip for the customer
    5. 🎯 Recommend the single BEST product and explain why
    
    Keep response friendly, helpful and concise.
    Use Indian rupee ₹ for prices.
    """
    
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[{"role": "user", "content": prompt}],
        max_tokens=600
    )
    
    return response.choices[0].message.content

print("AI Ranking function ready!")
```

```python
def smart_search(query, max_price=None, min_price=None,
                 category=None, top_n=5):
    
    print("=" * 55)
    print(f"  🔍 Smart Search: '{query}'")
    print("=" * 55)
    
    # Step 1: Get TF-IDF results
    cleaned_query = clean_query(query)
    query_vector  = vectorizer.transform([cleaned_query])
    scores        = cosine_similarity(query_vector, tfidf_matrix).flatten()
    
    temp_df                     = df.copy()
    temp_df['similarity_score'] = scores
    
    # Step 2: Apply filters
    if max_price:
        temp_df = temp_df[temp_df['discounted_price'] <= max_price]
        print(f"  💰 Max Price Filter : ₹{max_price:,}")
        
    if min_price:
        temp_df = temp_df[temp_df['discounted_price'] >= min_price]
        print(f"  💰 Min Price Filter : ₹{min_price:,}")
        
    if category:
        temp_df = temp_df[temp_df['category'].str.lower() == category.lower()]
        print(f"  📦 Category Filter  : {category}")
    
    # Step 3: Sort by similarity
    temp_df = temp_df.sort_values('similarity_score', ascending=False).head(top_n)
    
    if len(temp_df) == 0:
        print("\n❌ No products found! Try different filters.")
        return
    
    # Step 4: Show raw results
    print(f"\n📦 Top {len(temp_df)} Products Found:\n")
    display_cols = ['product_name', 'brand', 'category',
                    'discounted_price', 'rating',
                    'stock_status', 'similarity_score']
    print(temp_df[display_cols].to_string(index=False))
    
    # Step 5: Get AI ranking
    print("\n🤖 AI Analysis:\n")
    ai_response = ai_rank_products(query, temp_df)
    print(ai_response)
    print("=" * 55)

print("Smart Search function ready!")
```
